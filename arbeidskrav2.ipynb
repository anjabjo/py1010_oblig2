#### Oppgave 2
Det skal arrangeres en klassefest og man antar at hver elev spiser 1/4 pizza. Lag et program som tar inn antall elever fra konsollen ved

antall_elever = int(input('Skriv inn antall elever:' ))

Programmet skal så regne ut hvor mange pizzaer som skal handles inn til festen og skrive svaret til skjerm. Merk, man kan ikke kjøpe 4 og en kvart pizza på butikken (man må da kjøpe 5).

Hint1: Gir programmet ditt et fornuftig svar hvis det f.eks er 21 elever i klassen?

Hint2: Det er ikke vanlig å si/skrive: ‘Det må handles inn 6.0 pizzaer til festen’. Hvordan kan sikre at antall pizzaer skrives ut som et heltall (ikke desimaltall)?

```python
# Svar

""" 
Class party pizza calculator
By Anja Bjølgerud (abj@usn.no)
Updated 12.11.2026
"""

#%% Input
antall_elever = int(input('Skriv inn antall elever:' ))

#%% Calculation
hele, ekstra = divmod(antall_elever, 4)

if (ekstra == 0):
    antall_pizzaer = hele
else:
    antall_pizzaer = hele + 1

#%% Print
print(f"Det må bestilles {antall_pizzaer} pizzaer")
```

#### Oppgave 3

Resultatet v_rad skrives til skjerm med passende tekst og verdi.

Merk: np.pi er en ferdiglaget funksjon som gir verdien 3.1415....

```python
# Svar

""" 
Degrees to radians converter
By Anja Bjølgerud (abj@usn.no)
Updated 12.11.2026
"""

#%% Importing package
import numpy as np

#%% Input
v_grad = float(input('Skriv inn gradtallet:' ))

#%% Calculation
v_rad = v_grad*np.pi/180

#%% Print
print(f"{v_grad} grader tilsvarer {round(v_rad, 2)} radianer")
```

#### Oppgave 5
Lag et program med en funksjon som tar a og b som inn-argumenter og som så regner ut arealet og «ytre» omkrets til en figur satt sammen av en rettvinklet trekant og en halvsirkel, se figuren under. Med «ytre» omkrets menes samlet lengde av de sorte strekene. Funksjonen skal returnere arealet og «ytre» omkrets, som så skrives til skjerm med passende tekst.



```python
# Svar

""" 
Area and perimeter calculator
By Anja Bjølgerud (abj@usn.no)
Updated 12.11.2026
"""

#%% Importing package
import math

#%% Input
a = float(input("Skriv inn lengde a i cm"))
b = float(input("Skriv inn lengde b i cm"))

#%% Calculation of the hypothenus
c = math.sqrt(a**2 + b**2)   #Bruker Pythagoras til å beregne hypotenusen i den rettvinklede trekanten

#%% Calculation of area
r = a/2  # radius i sirkelen
areal_trekant = (a*b)/2
areal_halvsirkel = (math.pi * r**2)/2
areal_total = areal_trekant + areal_halvsirkel

#%% Calculation of perimeter
omkrets_halvsirkel = (2 * math.pi * r) / 2
omkrets_total = omkrets_halvsirkel + b + c

#%% Print
print(f"Når a = {a} cm og b = {b} cm blir arealet til figuren i oppgave 5: {round(areal_total,2)} cm² og omkretsen {round(omkrets_total,2)} cm.")


# kilde : https://www.w3schools.com/python/module_math.asp
```
